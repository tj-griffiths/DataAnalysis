## 1. Project Description

The report simulates making a measurement of a particle decay, from which you can measure a parameter related to the matter/antimatter asymmetry of the universe.

The signal decay $B \to DX$ follows the PDF:

$$p(t;\, \tau,\, \Delta m,\, V) \propto \left(1 + V \sin(\Delta m \times t)\right) \times \exp\left(-\frac{t}{\tau}\right) \tag{1}$$

where:
- $t$ is the quantity you observe, the decay time
- $\tau$ is a lifetime parameter
- $\Delta m$ is a parameter that leads to sinusoidal oscillations superimposed on the exponential decay
- $V$ is a parameter which measures the matter/antimatter asymmetry, with a value of $0$ if the universe is symmetric

The nominal values of the parameters are:
- $\tau = 1.5 \times 10^{-6}\ \text{s}$
- $\Delta m = 20.0 \times 10^{6}\ \text{s}^{-1}$
- $V = 0.1$

The detector only records events in the range $0.5 \times 10^{-6}\ \text{s} < t < 10 \times 10^{-6}\ \text{s}$. Note that equation $(1)$ is not yet normalised — this is something you will need to account for.

## 1.2 Step 2 

We now investigate the effect of a background component. Assume the background follows:

$$p(t;\, \sigma) \propto \exp\left(-\frac{t^2}{2\sigma^2}\right) \tag{2}$$

where $\sigma = 0.5 \times 10^{-6}\ \text{s}$.

You should:

- Generate pseudoexperiments where the background makes up **1%, 10%, or 20%** of the overall 10,000-event sample. Repeat the study for each case. *Remember the detector acceptance range applies here too.*
- For each background fraction, fit **with and without** the background component in the model and compare:
  - The **precision** obtained on each parameter
  - The **bias** on each parameter and how well it is known
  - Provide labelled plots demonstrating your conclusions
- You may treat $\sigma$ as known and fixed in the fit, but treat the **background fraction as a free parameter**.

> **Hints:** Generate events from both signal and background, then fit once with signal-only and once with signal+background PDFs. Normalise all PDFs before fitting. If a fit fails to converge, report this in your notebook.
